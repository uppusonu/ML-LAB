```python
import pandas as pd
from sklearn.preprocessing  import StandardScaler
```

```python
df = pd.read_csv("/content/placement_predict_50k_adjusted.csv")
```

```python
x = df[['CodingTestScore', 'AptitudeTestScore']]
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
scaled_df = pd.DataFrame(x_scaled, columns=['CodingTestScore', 'AptitudeTestScore'])
```
